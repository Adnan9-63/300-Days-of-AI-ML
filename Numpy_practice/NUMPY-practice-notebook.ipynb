 17. What is the result of the following expression? (★☆☆)

0 * np.nan
np.nan == np.nan
np.inf > np.nan
np.nan - np.nan
np.nan in set([np.nan])
0.3 == 3 * 0.1


```python
print(0 * np.nan)
print(np.nan == np.nan)
print(np.inf > np.nan)
print(np.nan - np.nan)
print(np.nan in set([np.nan]))
print(0.3 == 3 * 0.1)

'''

0 * np.nan---->>Nan
nanExplanation: Any arithmetic operation involving NaN (Not a Number) propagates NaN.

np.nan == np.nan ---->>False
Explanation: By the IEEE 754 floating-point standard, NaN is not equal to anything, including itself.


np.inf > np.nan ----->> False
Explanation: Any relational comparison (>, <, >=, <=) involving NaN evaluates to False

np.nan - np.nan------>> nan
Explanation: Similar to multiplication, subtraction with NaN results in NaN


np.nan in set([np.nan]) ------>> True
Explanation: Even though np.nan == np.nan is False, 
Python sets check object identity/hashing during lookup,
 allowing the identical object stored in the set to be found.


 0.3 == 3 * 0.1 ------>> False
 Explanation: Due to the limitations of binary floating-point representation in computers,
   3 * 0.1 evaluates to approximately 0.30000000000000004 rather than exact 0.3.
'''
```

 19. Create a 8x8 matrix and fill it with a checkerboard pattern (★☆☆)

```python
Z = np.zeros((8,8),dtype=int)
Z[1::2,::2] = 1
Z[::2,1::2] = 1
print(Z)
```
