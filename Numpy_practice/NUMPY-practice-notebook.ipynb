4. How to find the memory size of any array (★☆☆)

```python
array=np.ones((10,5))
print("Array of ones-->>",array)
size=array.size
items=array.itemsize
print("Size of array-->>",size)
print("Item size of array-->>",items)
Z = np.zeros((10,10))
print("%d bytes" % (Z.size * Z.itemsize))
# Simpler alternative
print("%d bytes" % Z.nbytes)
```
